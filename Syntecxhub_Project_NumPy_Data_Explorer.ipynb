```python
import numpy as np

print("NumPy Data Explorer")
```

```python
data = np.genfromtxt(
    "Star_Wars_Characters.csv",
    delimiter=";",
    dtype=str,
    encoding="cp1252",
    skip_header=1
)

print(data)
```

```python
print("Number of rows:", data.shape[0])
print("Number of columns:", data.shape[1])
print("Array shape:", data.shape)
print("Array size:", data.size)
```

```python
# Indexing

print("First character:")
print(data[0])

print("\nSecond character:")
print(data[1])

print("\nLast character:")
print(data[-1])
```

```python
# Access individual values

print("First character name:", data[0, 0])
print("First character birthplace:", data[0, 1])
print("First character age:", data[0, 2])
print("First character affiliation:", data[0, 3])
print("First character status:", data[0, 4])
```

```python
# Slicing

print("First 5 rows:")
print(data[:5])

print("\nFirst 5 character names:")
print(data[:5, 0])

print("\nFirst 10 character names and ages:")
print(data[:10, [0, 2]])
```

```python
# Extract Age column

age_text = data[:, 2]

print("Age column:")
print(age_text)
```

```python
# Clean the Age column

age_values = []

for age in age_text:
    age = age.strip()

    # Remove "~"
    age = age.replace("~", "")

    # Handle ranges such as 30–40
    if "–" in age:
        parts = age.split("–")
        try:
            first = float(parts[0])
            second = float(parts[1])
            age_values.append((first + second) / 2)
        except:
            age_values.append(np.nan)

    # Handle values such as "10 (20 biological)"
    elif "(" in age:
        try:
            age_values.append(float(age.split("(")[0].strip()))
        except:
            age_values.append(np.nan)

    # Handle "80+"
    elif "+" in age:
        try:
            age_values.append(float(age.replace("+", "")))
        except:
            age_values.append(np.nan)

    # Handle normal numbers
    else:
        try:
            age_values.append(float(age))
        except:
            age_values.append(np.nan)

age_values = np.array(age_values)

print("Cleaned age values:")
print(age_values)
```

```python
# Mathematical operations

valid_ages = age_values[~np.isnan(age_values)]

print("Sum of ages:", np.sum(valid_ages))
print("Minimum age:", np.min(valid_ages))
print("Maximum age:", np.max(valid_ages))
print("Average age:", np.mean(valid_ages))
```

```python
# Element-wise mathematical operations

print("First 10 ages:", valid_ages[:10])

print("\nAges + 5:")
print(valid_ages[:10] + 5)

print("\nAges × 2:")
print(valid_ages[:10] * 2)

print("\nAges squared:")
print(valid_ages[:10] ** 2)
```

```python
# Create a 2D age array for axis-wise operations

age_matrix = valid_ages[:60].reshape(3, 20)

print("Age matrix:")
print(age_matrix)

print("\nSum across rows (axis=1):")
print(np.sum(age_matrix, axis=1))

print("\nSum across columns (axis=0):")
print(np.sum(age_matrix, axis=0))
```

```python
# Statistical operations

print("Mean age:", np.mean(valid_ages))
print("Median age:", np.median(valid_ages))
print("Standard deviation:", np.std(valid_ages))
print("Variance:", np.var(valid_ages))
```

```python
# Reshaping

print("Original shape:", age_matrix.shape)

reshaped_age_matrix = age_matrix.reshape(4, 15)

print("New shape:", reshaped_age_matrix.shape)
print("\nReshaped age matrix:")
print(reshaped_age_matrix)
```

```python
# Broadcasting

print("Original first row:")
print(reshaped_age_matrix[0])

print("\nAfter adding 5 to every age:")
print(reshaped_age_matrix[0] + 5)

print("\nAfter multiplying every age by 2:")
print(reshaped_age_matrix[0] * 2)
```

```python
# Save NumPy array

np.save("star_wars_ages.npy", valid_ages)

print("NumPy array saved successfully.")
```

```python
# Load the saved NumPy array

loaded_ages = np.load("star_wars_ages.npy")

print("Loaded NumPy array:")
print(loaded_ages)

print("\nArray shape:", loaded_ages.shape)
```

```python
# Performance comparison: Python list vs NumPy array

import time

# Create larger datasets for a meaningful comparison
python_list = valid_ages.tolist() * 10000
numpy_array = np.tile(valid_ages, 10000)

# Python list operation
start = time.perf_counter()

python_result = [x * 2 for x in python_list]

python_time = time.perf_counter() - start

# NumPy array operation
start = time.perf_counter()

numpy_result = numpy_array * 2

numpy_time = time.perf_counter() - start

print("Number of elements:", len(python_list))
print("Python list time:", python_time, "seconds")
print("NumPy array time:", numpy_time, "seconds")

print("NumPy speedup:", python_time / numpy_time, "times")
```

```python
# Verify saved and loaded arrays

print("Original array shape:", valid_ages.shape)
print("Loaded array shape:", loaded_ages.shape)

print("Arrays are equal:", np.array_equal(valid_ages, loaded_ages))
```
